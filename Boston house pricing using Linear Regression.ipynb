### 🔍 Complete Case Analysis (CCA)

Complete Case Analysis is a missing-value handling technique in which all rows containing at least one missing value are removed.

Since the proportion of missing values in the affected features is relatively low, we will evaluate whether removing incomplete rows is appropriate for this dataset.

```python
#total rows before cca
print(df.shape)
#count rows containing at least one missing values
print(df.isnull().any(axis = 1).sum())
# percentage of rows that would be removed
print((df.isnull().any(axis=1).mean()*100).round(2))
```

##  6. Handling Missing Values Using Imputation

Imputation replaces missing values with estimated values rather than deleting entire rows.

In this project:
- **Numerical features:** Median imputation will be used as a baseline because the median is less sensitive to outliers.
- **Binary feature (`CHAS`):** Most-frequent-value imputation will be considered.

This approach preserves the dataset's observations while allowing us to prepare the data for regression modelling.

```python
from sklearn.impute import SimpleImputer

dataset_imputed = df.copy()

# Numerical features: median imputation
numerical_cols = ['CRIM', 'ZN', 'INDUS', 'AGE', 'LSTAT']

median_imputer = SimpleImputer(strategy='median')
dataset_imputed[numerical_cols] = median_imputer.fit_transform(
    dataset_imputed[numerical_cols]
)
# Binary feature: most frequent value
mode_imputer = SimpleImputer(strategy='most_frequent')
dataset_imputed[['CHAS']] = mode_imputer.fit_transform(
    dataset_imputed[['CHAS']]
)
```

```python
print("Missing values remaining:", dataset_imputed.isnull().sum())
print("Dataset shape:", dataset_imputed.shape)
```

### Performance metrics

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
print(mean_absolute_error(y_test, reg_pred))
print(mean_squared_error(y_test, reg_pred))
print(np.sqrt(mean_squared_error(y_test, reg_pred)))
```

### R score and adjusted R score 
where :
R^2 = 1 - SSR/SST
R^2 = Coefficient of determination SSR = sum of squares of residuals SST = total sum of squares

```python
from sklearn.metrics import r2_score
score = r2_score(y_test, reg_pred)
print(score)
```

### Adjusted R2 = 1-[(1-R^2)*(n-1)/(n-k-1)]
where: 
n = no of observations
k = independent features

```python
1-(1-score)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```
